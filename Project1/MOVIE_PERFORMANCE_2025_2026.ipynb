## Outlier

```python
import matplotlib.pyplot as plt

numerical_columns = [
    'rating_out_of_10',
    'vote_count',
    'runtime_minutes',
    'box_office_usd',
    'budget_usd'
]

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    print(f"\nKolom: {column}")
    print(f"Lower Bound: {lower_bound}")
    print(f"Upper Bound: {upper_bound}")
    print(f"Jumlah Outlier: {len(outliers)}")
```

```python
Q1 = df['box_office_usd'].quantile(0.25)
Q3 = df['box_office_usd'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_box_office = df[
    (df['box_office_usd'] < lower_bound) |
    (df['box_office_usd'] > upper_bound)
]

outlier_box_office[['title', 'box_office_usd']].sort_values(
    by='box_office_usd',
    ascending=False
).head(10)
```

```python
Q1 = df['budget_usd'].quantile(0.25)
Q3 = df['budget_usd'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_budget = df[
    (df['budget_usd'] < lower_bound) |
    (df['budget_usd'] > upper_bound)
]

outlier_budget[['title', 'budget_usd']].sort_values(
    by='budget_usd',
    ascending=False
).head(10)
```

# EDA

## ROI & PROFIT

```python
# menghitung profit
df['profit'] = df['box_office_usd'] - df['budget_usd']

# menghitung ROI
df['ROI'] = (
    (df['box_office_usd'] - df['budget_usd'])
    / df['budget_usd']
) * 100
```

```python
df.head(5)
```

## 1

1. Film dengan box office tertinggi

```python
top_10_revanue = df.sort_values(
    by='box_office_usd',
    ascending=False
).head(10)

print(top_10_revanue[
    ['title', 'box_office_usd', 'production_company']
])
```

## 2

2. Pendapatan tertinggi

```python
#Genre yang menghasilkan pendapatan tertinggi

genre_df = df.copy()
genre_df['genres'] = genre_df['genres'].str.split(', ')
genre_df = genre_df.explode('genres')

genre_revenue = (
    genre_df.groupby('genres')['box_office_usd'].sum().sort_values(ascending=False)

)
print(genre_revenue)
```

```python
# Tahun dengan pendapatan film terbesar
year_revenue = (
    df.groupby('release_year')['box_office_usd'].sum().sort_values(ascending=False)
)
print(year_revenue)
```

```python
# Bulan dengan pendapatan film terbesar
month__avg_revenue = (
    df.groupby('release_month')['box_office_usd'].mean().sort_values(ascending=False)
)
print(month__avg_revenue)
```

```python
genre_rating = (
    genre_df.groupby('genres')['rating_out_of_10'].mean().sort_values(ascending=False)
)

print(genre_rating.head(10))
```

```python
correlation = df[
    ['rating_out_of_10', 'box_office_usd']
].corr()

print("correlation:", correlation)
```

Berdasarkan analisis korelasi, hubungan antara rating film dan pendapatan box office menunjukkan nilai korelasi sebesar 0.0028, yang sangat mendekati nol. Hal ini menunjukkan bahwa rating film tidak memiliki hubungan linear yang signifikan dengan pendapatan box office pada dataset ini. Dengan demikian, rating yang tinggi belum tentu menghasilkan pendapatan yang tinggi.

4. Budget vs box office

Berdasarkan analisis korelasi, hubungan antara rating film dan pendapatan box office menunjukkan nilai korelasi sebesar 0.9735, yang sangat mendekati 1. Hal ini menunjukkan bahwa budget lebih tinggi menghasilkan box office lebih tinggi.

## 6

6. Studio dengan penghasilan revenue terbesar

```python
studio_df = df.copy()

studio_df['production_company'] = (
    studio_df['production_company']
    .str.split('/')
)

studio_df = studio_df.explode('production_company')

studio_df['production_company'] = (
    studio_df['production_company']
    .str.strip()
)
```

```python
studio_revenue = (
    studio_df.groupby('production_company')['box_office_usd']
    .sum()
    .sort_values(ascending=False)
)

print(studio_revenue.head(10))
```

```python
studio_analysis = (
    studio_df.groupby('production_company')
    .agg(
        average_rating=('rating_out_of_10', 'mean'),
        movie_count=('title', 'count')
    )
)

studio_analysis = studio_analysis[
    studio_analysis['movie_count'] >= 5
]

studio_analysis = studio_analysis.sort_values(
    by='average_rating',
    ascending=False
)

print(studio_analysis.head(10))
```

```python
import sqlite3

# Membuat database sementara di Colab
conn = sqlite3.connect(":memory:")

# Memasukkan dataframe menjadi tabel SQL bernama movies
df.to_sql("movies", conn, index=False, if_exists="replace")
```

```python
query = """
SELECT *
FROM movies
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT COUNT(*) AS total_movie
FROM movies;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT genres,
       COUNT(*) AS total_movie
FROM movies
GROUP BY genres
ORDER BY total_movie DESC;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT title,
       box_office_usd
FROM movies
ORDER BY box_office_usd DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT genres,
       ROUND(AVG(rating_out_of_10),2) AS avg_rating
FROM movies
GROUP BY genres
ORDER BY avg_rating DESC;
"""

pd.read_sql(query, conn)
```
